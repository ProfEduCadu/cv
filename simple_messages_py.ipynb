O !r deixa \t e \n visíveis na saída. lstrip() limpa o início, rstrip() limpa o fim e strip() limpa os dois lados.

```python
name = "\t\n Cadu \n\t"

print(f"Original: {name!r}")
print(f"lstrip(): {name.lstrip()!r}")
print(f"rstrip(): {name.rstrip()!r}")
print(f"strip(): {name.strip()!r}")
```
