## Como funciona

O programa apresenta três questões, confere as respostas e mostra a pontuação. Os assuntos das questões erradas aparecem ao final como sugestão de revisão.

## Como usar

Execute a célula de código, digite cada resposta no campo exibido e pressione Enter. Na questão de frações, você pode responder `0,75` ou `0.75`.

```python
# FLASHMATH — versão 1

atividades = [
    {
        "assunto": "Porcentagem",
        "pergunta": "Quanto é 25% de 80?",
        "resposta": 20
    },
    {
        "assunto": "Frações",
        "pergunta": "Quanto é 1/2 + 1/4? Responda em decimal.",
        "resposta": 0.75
    },
    {
        "assunto": "Regras de sinais",
        "pergunta": "Quanto é (-3) × (-4)?",
        "resposta": 12
    }
]

acertos = 0
assuntos_para_revisar = []

print("=== FLASHMATH ===")

for numero, atividade in enumerate(atividades, start=1):
    print(f"\nAtividade {numero} — {atividade['assunto']}")
    print(atividade["pergunta"])

    try:
        texto = input("Sua resposta: ")
        resposta_aluno = float(texto.replace(",", "."))

        if abs(resposta_aluno - atividade["resposta"]) < 0.000001:
            print("✅ Correto!")
            acertos += 1
        else:
            print(f"❌ Resposta correta: {atividade['resposta']}")
            assuntos_para_revisar.append(atividade["assunto"])

    except ValueError:
        print("❌ Digite um número. Esta questão ficará para revisão.")
        assuntos_para_revisar.append(atividade["assunto"])

print("\n=== RESULTADO ===")
print(f"Acertos: {acertos} de {len(atividades)}")

if assuntos_para_revisar:
    print("Assuntos para revisar:")
    for assunto in assuntos_para_revisar:
        print(f"- {assunto}")
else:
    print("Parabéns! Você acertou todas as atividades.")
```
