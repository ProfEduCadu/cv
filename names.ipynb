A lista se chama names. Em Python, as posições começam em 0: names[0] acessa o primeiro nome. Troque os exemplos pelos nomes dos seus amigos.

```python
names = ["Ana", "Bruno", "Carlos", "Carla"]

print(names[0])
print(names[1])
print(names[2])
print(names[3])
```

A lista guests guarda os três convidados. Cada print() usa uma posição da lista para personalizar o convite. Você pode trocar os nomes por pessoas que gostaria de convidar.

```python
guests = ["Ada Lovelace", "Albert Einstein", "Paulo Freire"]

print(f"Dear {guests[0]}, I would like to invite you to dinner.")
print(f"Dear {guests[1]}, I would like to invite you to dinner.")
print(f"Dear {guests[2]}, I would like to invite you to dinner.")
```

A linha guests[1] = "Marie Curie" substitui o segundo nome da lista.

```python
guests = ["Ada Lovelace", "Albert Einstein", "Paulo Freire"]

# Convites originais
print(f"Dear {guests[0]}, I would like to invite you to dinner.")
print(f"Dear {guests[1]}, I would like to invite you to dinner.")
print(f"Dear {guests[2]}, I would like to invite you to dinner.")

# Aviso e troca de convidado
print(f"\n{guests[1]} can't make it to dinner.")
guests[1] = "Marie Curie"

# Novos convites
print(f"\nDear {guests[0]}, I would like to invite you to dinner.")
print(f"Dear {guests[1]}, I would like to invite you to dinner.")
print(f"Dear {guests[2]}, I would like to invite you to dinner.")
```

Ao final, guests terá seis pessoas. insert(posição, nome) coloca alguém na posição escolhida; append(nome) acrescenta ao fim.

```python
guests = ["Ada Lovelace", "Marie Curie", "Paulo Freire"]

print("I found a bigger dinner table!")

guests.insert(0, "Katherine Johnson")  # Início
guests.insert(2, "Alan Turing")         # Meio
guests.append("Malala Yousafzai")       # Fim

print(f"Dear {guests[0]}, I would like to invite you to dinner.")
print(f"Dear {guests[1]}, I would like to invite you to dinner.")
print(f"Dear {guests[2]}, I would like to invite you to dinner.")
print(f"Dear {guests[3]}, I would like to invite you to dinner.")
print(f"Dear {guests[4]}, I would like to invite you to dinner.")
print(f"Dear {guests[5]}, I would like to invite you to dinner.")
```

pop() retira o último nome e permite usá-lo na mensagem de desculpas. No final, [] confirma que a lista ficou vazia.

```python
guests = [
    "Katherine Johnson",
    "Ada Lovelace",
    "Alan Turing",
    "Marie Curie",
    "Paulo Freire",
    "Malala Yousafzai",
]

print("I'm sorry, but I can invite only two people to dinner.")

removed_guest = guests.pop()
print(f"Dear {removed_guest}, I'm sorry I can't invite you to dinner.")

removed_guest = guests.pop()
print(f"Dear {removed_guest}, I'm sorry I can't invite you to dinner.")

removed_guest = guests.pop()
print(f"Dear {removed_guest}, I'm sorry I can't invite you to dinner.")

removed_guest = guests.pop()
print(f"Dear {removed_guest}, I'm sorry I can't invite you to dinner.")

print(f"Dear {guests[0]}, you're still invited to dinner!")
print(f"Dear {guests[1]}, you're still invited to dinner!")

del guests[1]
del guests[0]

print(guests)  # Resultado: []
```
