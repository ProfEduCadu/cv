# Analisador de notas de uma turma

Projeto desenvolvido por Prof. Cadu para praticar Python e visualização de dados.

## Objetivo

Calcular a média, a maior e a menor nota de uma turma e mostrar as notas em um gráfico de barras.

## Dados

Os nomes e as notas usados neste exemplo são fictícios.

## Como usar

Execute as células de código de cima para baixo. Para analisar outra turma, altere os nomes na lista `alunos` e os valores na lista `notas`, mantendo uma nota para cada aluno.

```python
alunos = ["Ana", "Bruno", "Carla", "Diego"]
notas = [8.0, 5.0, 9.0, 6.5]

media_turma = sum(notas) / len(notas)

print("NOTAS DA TURMA")
for aluno, nota in zip(alunos, notas):
  print(f"{aluno}:{nota:.1f}")

print(f"Média da turma: {media_turma:.1f}")
print(f"Maior nota: {max(notas):.1f}")
print(f"Menor nota: {min(notas):.1f}")
```

Criar um gráfico

```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt

cores = ["tomato" if nota < 6 else "steelblue" for nota in notas]

plt.bar(alunos, notas, color=cores)
plt.axhline(6, color="red", linestyle="--", label="Nota 6")
plt.ylim(0, 10)
plt.ylabel("Nota")
plt.title("Notas da turma")
plt.legend()
plt.show()
```
