Diferença entre os comandos:
Comando       O que faz       Modifica a lista original?
`sorted(places)` Retorna uma nova lista em ordem alfabética | Não |

`sorted(places, reverse=True)` | Retorna uma nova lista em ordem alfabética inversa | Não |

`places.reverse()` | Inverte a ordem atual dos elementos | Sim |

`places.sort()` | Ordena os elementos alfabeticamente | Sim |

`places.sort(reverse=True)` | Ordena os elementos alfabeticamente de trás para frente | Sim |

```python
# Cinco lugares fora da ordem alfabética
places = ["Tokyo", "Paris", "London", "Cairo", "Amsterdam"]

# Lista na ordem original
print("Ordem original:")
print(places)

# Ordem alfabética sem modificar a lista
print("\nOrdem alfabética com sorted():")
print(sorted(places))

# Confirma que a lista continua na ordem original
print("\nLista original preservada:")
print(places)

# Ordem alfabética inversa sem modificar a lista
print("\nOrdem alfabética inversa com sorted():")
print(sorted(places, reverse=True))

# Confirma novamente a ordem original
print("\nLista original preservada:")
print(places)

# Inverte a ordem atual da lista
places.reverse()
print("\nLista após reverse():")
print(places)

# Inverte novamente, recuperando a ordem original
places.reverse()
print("\nDe volta à ordem original:")
print(places)

# Modifica a lista para a ordem alfabética
places.sort()
print("\nLista após sort():")
print(places)

# Modifica a lista para a ordem alfabética inversa
places.sort(reverse=True)
print("\nLista após sort(reverse=True):")
print(places)
```

Observe que reverse() apenas inverte a sequência atual: ele não organiza os nomes alfabeticamente.

vamos criar uma lista de linguagens de programação e usar as funções e os métodos estudados. Execute no Colab ou salve como every_function.py:

```python
languages = ["Python", "Java", "C", "JavaScript", "Ruby"]

print("Lista original:")
print(languages)

# Acessar elementos
print("\nPrimeira linguagem:", languages[0])
print("Última linguagem:", languages[-1])

# Modificar um elemento
languages[1] = "Go"
print("\nApós substituir Java por Go:")
print(languages)

# append(): adicionar ao final
languages.append("Rust")
print("\nApós append():")
print(languages)

# insert(): adicionar em uma posição específica
languages.insert(1, "Java")
print("\nApós insert():")
print(languages)

# del: excluir pela posição
del languages[2]
print("\nApós excluir o elemento da posição 2:")
print(languages)

# pop(): remover e retornar o último elemento
removed_language = languages.pop()
print("\nLinguagem removida com pop():", removed_language)
print(languages)

# pop() também pode remover pela posição
removed_language = languages.pop(1)
print("\nLinguagem removida da posição 1:", removed_language)
print(languages)

# remove(): excluir pelo valor
languages.remove("Ruby")
print("\nApós remove('Ruby'):")
print(languages)

# sorted(): ordenar sem modificar a lista original
print("\nOrdem alfabética:")
print(sorted(languages))

print("\nOrdem alfabética inversa:")
print(sorted(languages, reverse=True))

print("\nLista original preservada:")
print(languages)

# reverse(): inverter a ordem atual
languages.reverse()
print("\nApós reverse():")
print(languages)

# sort(): modificar a lista para a ordem alfabética
languages.sort()
print("\nApós sort():")
print(languages)

# sort(reverse=True): ordenar em ordem alfabética inversa
languages.sort(reverse=True)
print("\nApós sort(reverse=True):")
print(languages)

# len(): contar os elementos
print(f"\nA lista contém {len(languages)} linguagens.")
```

Lembrete: as posições começam em 0. O comando del exclui um elemento; pop() permite guardar o elemento removido em uma variável.
